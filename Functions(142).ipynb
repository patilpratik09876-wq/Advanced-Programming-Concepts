```python
# Program 1: Factorial of a Number

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact

n = int(input("Enter a number: "))
print("Factorial =", factorial(n))
```

```python
# Program 3: Calculate Simple Interest

def simple_interest(p, r, t):
    si = (p * r * t) / 100
    return si

p = float(input("Enter principal amount: "))
r = float(input("Enter rate of interest: "))
t = float(input("Enter time: "))

print("Simple Interest =", simple_interest(p, r, t))
```

```python
# Program 4: Check Prime Number

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

n = int(input("Enter a number: "))
print(is_prime(n))
```

```python
# Program 14: Find Unique Elements in a List

def unique_elements(lst):
    unique = []

    for x in lst:
        if x not in unique:
            unique.append(x)

    return unique

lst = list(map(int, input("Enter list elements: ").split()))
print("Unique Elements =", unique_elements(lst))
```

```python
# Program 15: Find Second Largest Number

def second_largest(lst):
    unique = []

    for x in lst:
        if x not in unique:
            unique.append(x)

    unique.sort()
    return unique[-2]

lst = list(map(int, input("Enter numbers: ").split()))
print("Second Largest =", second_largest(lst))
```

```python
# Program 16: First N Fibonacci Numbers

def fibonacci(n):
    series = []
    a = 0
    b = 1

    for i in range(n):
        series.append(a)
        a, b = b, a + b

    return series

n = int(input("Enter number of terms: "))
print("Fibonacci Series =", fibonacci(n))
```

```python
# Program 17: Calculate Percentage and Grade

def result(m1, m2, m3, m4, m5):
    total = m1 + m2 + m3 + m4 + m5
    percentage = total / 5

    if percentage >= 75:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 50:
        grade = "C"
    elif percentage >= 40:
        grade = "D"
    else:
        grade = "F"

    return percentage, grade

marks = []
for i in range(5):
    marks.append(float(input("Enter marks: ")))

percentage, grade = result(*marks)
print("Percentage =", percentage)
print("Grade =", grade)
```

```python
# Program 20: Calculate Total Bill with Discount

def total_bill(prices, quantities):
    total = 0

    for i in range(len(prices)):
        total += prices[i] * quantities[i]

    if total >= 5000:
        discount = total * 0.10
    else:
        discount = 0

    return total - discount

n = int(input("Enter number of items: "))
prices = []
quantities = []

for i in range(n):
    prices.append(float(input("Enter price: ")))
    quantities.append(int(input("Enter quantity: ")))

print("Final Bill =", total_bill(prices, quantities))
```

```python
# Program 21: Find Minimum Maximum Sum and Average

def list_details(numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    total = sum(numbers)
    average = total / len(numbers)
    return minimum, maximum, total, average

numbers = list(map(int, input("Enter numbers: ").split()))
minimum, maximum, total, average = list_details(numbers)

print("Minimum =", minimum)
print("Maximum =", maximum)
print("Sum =", total)
print("Average =", average)
```

```python
# Program 22: Student Record Processing

def calculate_result(marks):
    total = sum(marks)
    percentage = total / 5

    if percentage >= 75:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 50:
        grade = "C"
    elif percentage >= 40:
        grade = "D"
    else:
        grade = "F"

    return total, percentage, grade

students = []
n = int(input("Enter number of students: "))

for i in range(n):
    name = input("Enter name: ")
    roll = int(input("Enter roll number: "))
    marks = []

    for j in range(5):
        marks.append(float(input("Enter marks: ")))

    total, percentage, grade = calculate_result(marks)

    students.append({
        "name": name,
        "roll": roll,
        "marks": marks,
        "total": total,
        "percentage": percentage,
        "grade": grade
    })

class_average = sum(s["percentage"] for s in students) / len(students)
highest = max(students, key=lambda x: x["percentage"])
lowest = min(students, key=lambda x: x["percentage"])

for s in students:
    print(s)

print("Class Average =", class_average)
print("Highest Scorer =", highest["name"])
print("Lowest Scorer =", lowest["name"])
```

```python
# Program 23: Simple Banking System

balance = 0
transactions = []

def deposit(amount):
    global balance
    balance += amount
    transactions.append("Deposited " + str(amount))

def withdraw(amount):
    global balance
    if amount <= balance:
        balance -= amount
        transactions.append("Withdrawn " + str(amount))
    else:
        print("Insufficient Balance")

def balance_enquiry():
    print("Balance =", balance)

def transaction_history():
    for transaction in transactions:
        print(transaction)

deposit(5000)
withdraw(2000)
balance_enquiry()
transaction_history()
```

```python
# Program 24: Library Management System

books = {}

def add_book(name):
    books[name] = True

def issue_book(name):
    if name in books and books[name]:
        books[name] = False
        print("Book Issued")
    else:
        print("Book Not Available")

def return_book(name):
    if name in books:
        books[name] = True
        print("Book Returned")

def search_book(name):
    if name in books:
        print("Book Found")
    else:
        print("Book Not Found")

def display_books():
    for book in books:
        if books[book]:
            print(book)

add_book("Python")
add_book("Java")
issue_book("Python")
return_book("Python")
search_book("Java")

print("Available Books:")
display_books()
```

```python
# Program 25: Modular Electricity Billing

def calculate_units(units):
    if units <= 100:
        return units * 2
    elif units <= 200:
        return 200 + (units - 100) * 3
    else:
        return 500 + (units - 200) * 5

def add_fixed_charge(amount):
    return amount + 100

def add_tax(amount):
    return amount + amount * 0.05

def apply_discount(amount):
    if amount > 1000:
        amount -= amount * 0.05
    return amount

units = int(input("Enter units consumed: "))
bill = calculate_units(units)
bill = add_fixed_charge(bill)
bill = add_tax(bill)
bill = apply_discount(bill)

print("Final Electricity Bill =", bill)
```

```python
# Program 26: Hospital Billing System

def consultation_charge():
    return float(input("Enter consultation charges: "))

def laboratory_charge():
    return float(input("Enter laboratory charges: "))

def medicine_charge():
    return float(input("Enter medicine charges: "))

def room_charge():
    return float(input("Enter room charges: "))

def final_bill(total, category):
    if category.lower() == "senior":
        total -= total * 0.10
    elif category.lower() == "child":
        total -= total * 0.05
    return total

total = consultation_charge()
total += laboratory_charge()
total += medicine_charge()
total += room_charge()

category = input("Enter patient category: ")
print("Final Bill =", final_bill(total, category))
```

```python
# Program 27: Shopping Invoice System

products = {}

def add_product(name, price, quantity):
    products[name] = [price, quantity]

def remove_product(name):
    if name in products:
        del products[name]

def calculate_subtotal():
    total = 0
    for value in products.values():
        total += value[0] * value[1]
    return total

def apply_coupon(total):
    coupon = input("Enter coupon code: ")
    if coupon == "SAVE10":
        total -= total * 0.10
    return total

def calculate_gst(total):
    return total + total * 0.18

def invoice():
    subtotal = calculate_subtotal()
    subtotal = apply_coupon(subtotal)
    total = calculate_gst(subtotal)
    print("Final Invoice =", total)

add_product("Laptop", 50000, 1)
add_product("Mouse", 500, 2)
invoice()
```

```python
# Program 28: Recursive Binary Search

def binary_search(lst, target, low, high):
    if low > high:
        return -1

    mid = (low + high) // 2

    if lst[mid] == target:
        return mid
    elif target < lst[mid]:
        return binary_search(lst, target, low, mid - 1)
    else:
        return binary_search(lst, target, mid + 1, high)

lst = list(map(int, input("Enter sorted numbers: ").split()))
target = int(input("Enter element to search: "))

result = binary_search(lst, target, 0, len(lst) - 1)

if result == -1:
    print("Element Not Found")
else:
    print("Element Found at Index", result)
```

```python
# Program 29: Decimal to Binary Using Recursion

def decimal_to_binary(n):
    if n == 0:
        return ""
    return decimal_to_binary(n // 2) + str(n % 2)

n = int(input("Enter decimal number: "))

if n == 0:
    print("Binary = 0")
else:
    print("Binary =", decimal_to_binary(n))
```

```python
# Program 30: Check Palindrome Using Recursion

def palindrome(text):
    if len(text) <= 1:
        return True

    if text[0] != text[-1]:
        return False

    return palindrome(text[1:-1])

text = input("Enter a string: ")

if palindrome(text):
    print("Palindrome")
else:
    print("Not Palindrome")
```

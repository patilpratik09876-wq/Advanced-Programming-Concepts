```python
# Upload CSV Files in Google Colab
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt
import io
uploaded = files.upload()
```

```python
# 1. Student Marks Bar Chart
students = pd.read_csv(io.BytesIO(uploaded['students.csv']))
students.set_index('Name')[['Python', 'DBMS', 'Maths']].plot(kind='bar', figsize=(9, 5))
plt.title('Student Subject Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2. Student Average Marks Horizontal Bar Chart
students = pd.read_csv(io.BytesIO(uploaded['students.csv']))
students['Average'] = students[['Python', 'DBMS', 'Maths']].mean(axis=1)
plt.barh(students['Name'], students['Average'])
plt.title('Average Marks of Students')
plt.xlabel('Average Marks')
plt.ylabel('Student')
plt.tight_layout()
plt.show()
```

```python
# 3. Subject-Wise Average Marks Bar Chart
students = pd.read_csv(io.BytesIO(uploaded['students.csv']))
averages = students[['Python', 'DBMS', 'Maths']].mean()
plt.bar(averages.index, averages.values)
plt.title('Subject-Wise Average Marks')
plt.xlabel('Subject')
plt.ylabel('Average Marks')
plt.show()
```

```python
# 4. Employee Salary Bar Chart
employees = pd.read_csv(io.BytesIO(uploaded['employees.csv']))
plt.bar(employees['Name'], employees['Salary'])
plt.title('Employee Salaries')
plt.xlabel('Employee')
plt.ylabel('Salary (Rs)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 5. Department-Wise Average Salary Bar Chart
employees = pd.read_csv(io.BytesIO(uploaded['employees.csv']))
averages = employees.groupby('Department')['Salary'].mean()
plt.bar(averages.index, averages.values)
plt.title('Department-Wise Average Salary')
plt.xlabel('Department')
plt.ylabel('Average Salary (Rs)')
plt.show()
```

```python
# 6. Employee Experience and Salary Scatter Plot
employees = pd.read_csv(io.BytesIO(uploaded['employees.csv']))
plt.scatter(employees['Experience'], employees['Salary'])
plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary (Rs)')
plt.show()
```

```python
# 7. Disease-Wise Patient Count Pie Chart
patients = pd.read_csv(io.BytesIO(uploaded['patients.csv']))
counts = patients['Disease'].value_counts()
plt.pie(counts.values, labels=counts.index, autopct='%1.1f%%')
plt.title('Patients by Disease')
plt.show()
```

```python
# 8. Patient Medical Expense Bar Chart
patients = pd.read_csv(io.BytesIO(uploaded['patients.csv']))
plt.bar(patients['Name'], patients['Medical_Expense'])
plt.title('Patient Medical Expenses')
plt.xlabel('Patient')
plt.ylabel('Medical Expense (Rs)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 9. City-Wise Daily Temperature Line Chart
weather = pd.read_csv(io.BytesIO(uploaded['weather.csv']))
for city, group in weather.groupby('City'):
    group = group.sort_values('Date')
    plt.plot(group['Date'], group['Temperature'], marker='o', label=city)
plt.title('Daily Temperature by City')
plt.xlabel('Date')
plt.ylabel('Temperature (C)')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 10. City-Wise Average Temperature Bar Chart
weather = pd.read_csv(io.BytesIO(uploaded['weather.csv']))
averages = weather.groupby('City')['Temperature'].mean()
plt.bar(averages.index, averages.values)
plt.title('Average Temperature by City')
plt.xlabel('City')
plt.ylabel('Average Temperature (C)')
plt.show()
```

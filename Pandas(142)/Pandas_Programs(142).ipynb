```python
# 1. Student DataFrame and Marks Analysis
import pandas as pd
d={'Student_ID':[1,2,3,4,5],'Name':['Amit','Sneha','Ravi','Priya','Neha'],'Python':[85,70,90,65,80],'DBMS':[78,75,92,68,83],'Mathematics':[88,72,87,70,79]}
df=pd.DataFrame(d)
df['Total']=df[['Python','DBMS','Mathematics']].sum(axis=1)
df['Average']=df[['Python','DBMS','Mathematics']].mean(axis=1)
print(df)
print('Above 75%:',df[df['Average']>75])
```

```python
# 2. Employee DataFrame Analysis
import pandas as pd
d={'Employee_ID':[1,2,3,4,5],'Name':['Amit','Riya','Raj','Pooja','Neha'],'Department':['CSE','HR','IT','CSE','Sales'],'Salary':[45000,55000,75000,60000,48000],'Experience':[2,4,8,6,3]}
df=pd.DataFrame(d)
print(df[df['Salary']>50000])
print('Average salary:',df['Salary'].mean())
print('Highest salary:',df['Salary'].max())
print('Most experienced:',df.loc[df['Experience'].idxmax()])
```

```python
# 3. Product Total Sales
import pandas as pd
d={'Product_ID':[1,2,3,4,5],'Product_Name':['Pen','Book','Bag','Mouse','Keyboard'],'Category':['Stationery','Stationery','Accessories','Electronics','Electronics'],'Price':[20,150,900,500,1200],'Quantity':[50,40,15,20,12]}
df=pd.DataFrame(d)
df['Total_Amount']=df['Price']*df['Quantity']
print(df)
print('Highest total sales:',df.loc[df['Total_Amount'].idxmax()])
```

```python
# 4. Patient DataFrame Analysis
import pandas as pd
d={'Patient_ID':[1,2,3,4,5],'Name':['Asha','Ravi','Sita','Raj','Mina'],'Age':[65,45,72,35,61],'Disease':['Fever','Diabetes','Heart Disease','Cold','Arthritis'],'Medical_Charges':[12000,45000,80000,5000,55000]}
df=pd.DataFrame(d)
print('Above 60:',df[df['Age']>60])
print('Average charge:',df['Medical_Charges'].mean())
print('Maximum charge:',df['Medical_Charges'].max())
print('Charges above 50000:',df[df['Medical_Charges']>50000])
```

```python
# 5. Order Final Amount
import pandas as pd
d={'Order_ID':[1,2,3,4,5],'Customer':['Amit','Neha','Ravi','Pooja','Raj'],'Product':['Phone','Book','Laptop','Bag','Tablet'],'Quantity':[1,4,1,2,1],'Price':[15000,500,55000,2500,18000],'Discount':[1000,100,5000,300,1500]}
df=pd.DataFrame(d)
df['Final_Amount']=df['Quantity']*df['Price']-df['Discount']
print('All orders:',df)
print('Above 5000:',df[df['Final_Amount']>5000])
print('Highest-value order:',df.loc[df['Final_Amount'].idxmax()])
print('Average order value:',df['Final_Amount'].mean())
```

```python
# 6. Student Attendance DataFrame
import pandas as pd
d={'Student_ID':[1,2,3,4,5],'Name':['Amit','Neha','Raj','Priya','Ravi'],'Department':['CSE','IT','CSE','ENTC','IT'],'Total_Classes':[100,100,100,100,100],'Classes_Attended':[80,68,90,72,75]}
df=pd.DataFrame(d)
df['Attendance_Percentage']=df['Classes_Attended']/df['Total_Classes']*100
print(df)
print('Below 75%:',df[df['Attendance_Percentage']<75])
```

```python
# 7. Retail Shop Sales DataFrame
import pandas as pd
d={'Product_ID':[1,2,3,4,5],'Product_Name':['Book','Bag','Phone','Pen','Mouse'],'Category':['Stationery','Accessories','Electronics','Stationery','Electronics'],'Price':[200,1200,15000,20,600],'Quantity':[40,10,3,100,20]}
df=pd.DataFrame(d)
df['Total_Sales']=df['Price']*df['Quantity']
print(df)
print('Sales above 10000:',df[df['Total_Sales']>10000])
print('Maximum sales:',df.loc[df['Total_Sales'].idxmax()])
print('Average sales:',df['Total_Sales'].mean())
```

```python
# 8. Student Marks Pandas Series
import pandas as pd
s=pd.Series({'Amit':85,'Neha':70,'Ravi':92,'Priya':78,'Raj':65})
print(s)
print('Amit marks:',s['Amit'])
print('Maximum:',s.max())
print('Minimum:',s.min())
print('Average:',s.mean())
print('Above 75:',s[s>75])
```

```python
# 9. Employee Salary Pandas Series
import pandas as pd
s=pd.Series({'Amit':45000,'Neha':60000,'Raj':72000,'Priya':48000,'Ravi':55000})
print(s)
print('Highest salary:',s.max())
print('Lowest salary:',s.min())
print('Average salary:',s.mean())
print('Above 50000:',s[s>50000])
```

```python
# 10. Product Prices Pandas Series
import pandas as pd
s=pd.Series({'Pen':20,'Book':200,'Bag':1500,'Mouse':800,'Keyboard':1200})
print('Original prices:',s)
s=s*1.10
print('After 10% increase:',s)
print('Most expensive:',s.idxmax(),s.max())
print('Above 1000:',s[s>1000])
```

```python
# 11. Patient Ages Pandas Series
import pandas as pd
s=pd.Series({101:65,102:42,103:78,104:25,105:61})
print(s)
print('Average age:',s.mean())
print('Oldest patient:',s.idxmax(),s.max())
print('Youngest patient:',s.idxmin(),s.min())
print('Above 60:',s[s>60])
```

```python
# 12. Student Attendance Pandas Series
import pandas as pd
s=pd.Series({'Amit':82,'Neha':70,'Ravi':95,'Priya':92,'Raj':65})
print(s)
print('Average attendance:',s.mean())
print('Below 75%:',s[s<75])
print('Above 90%:',s[s>90])
print('Highest attendance:',s.max())
```
